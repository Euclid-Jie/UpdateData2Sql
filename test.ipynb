## 获取ETF实时数据

```python
import requests
import json
import pandas as pd
url = "https://push2.eastmoney.com/api/qt/stock/details/get?"
params = {
    "fields1": "f1,f2,f3,f4",
    "fields2": "f51,f52,f53,f54,f55",
    "fltt": "2",
    "cb": "jQuery35106793092372227159_1760358891711",
    "pos": "-11",
    "secid": "105.NVDA",
    "ut": "fa5fd1943c7b386f172d6893dbfba10b",
    "wbp2u": "|0|0|0|web",
}
```

```python
r = requests.get(url, params=params)
# 去除 JSONP 回调包装，提取 JSON 内容
response_text = r.text
start = response_text.find('(') + 1
end = response_text.rfind(')')
json_str = response_text[start:end]
data = json.loads(json_str)

temp_df = pd.DataFrame([item.split(",") for item in data["data"]["details"]])
temp_df
```

# wind指数网站获取指数数据

```python
import requests
import pandas as pd

# https://www.windindices.com/indices/zh/IndexF9/e2d5a98547c3ee7c923a0259cee963e4
url = "https://indexapi.wind.com.cn/indicesWebsite/api/Kline?indexId=6644c422b6edae80b3c7a7d55803bc9e&period=1Y&lan=cn"
res = requests.get(url)
data = pd.DataFrame(res.json()["Result"]["data"])
data = data[["tradeDate", "open", "hight", "low", "close", "pctChange", "volume", "amount"]]
```
